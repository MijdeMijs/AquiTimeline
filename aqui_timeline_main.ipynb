# Timeline

1. Set the academic year. If `set_academic_year` is not run: default = [2025, 2026].

```python
at.set_academic_year([2025, 2026])
```

2. Run `Event.reset_events()` to clear all events and prevent duplicates.
3. Create events: `Event(name="str", event_type="str", month=int, day=int)`
   
   Options for `Event()`:
    * `name`: any string;
    * `event_type`: `competition` or `activity` or `exam` (exam week) or `resit` (resit week) or `other` (other activity, e.g. afzwemmen)*;
    * `month`: integer (1 to 12);
    * `day`: integer (1 to 31, depending on the month).
4. Plot the timeline with `plot_timeline()`. `Event.all_events` serves as input to enter all created events.

<small>*If another string is given in `event_type`: by default, this will appear black on the timeline. With `set_colors()`, the color for this alternative event can be color mapped (see Support). The marker is always of the short type.</small>  

```python
# Reset events
at.Event.reset_events()

# Create events:

  # Example
at.Event(name="example_activity", event_type="activity", month=1, day=1)
at.Event(name="example_competition", event_type="competition", month=2, day=1)
at.Event(name="example_other", event_type="other", month=3, day=1)
at.Event(name="example_exam", event_type="exam", month=4, day=1)
at.Event(name="example_resit", event_type="resit", month=5, day=1)
at.Event(name="example_holiday", event_type="holiday", month=7, day=1)

at.Event(name="example_other_string", event_type="other_string", month=10, day=1)

# Plot timeline
at.plot_timeline(at.Event.all_events)
```
